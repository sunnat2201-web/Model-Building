```python
def toldir(df):
        
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='str':
                df.fillna({col:df[col].mode()[0]},inplace=True)
            else:
                df.fillna({col:df[col].mean()},inplace=True)
                
    return(df)
```

```python
toldir(df)
```

```python

df=toldir(df)
df.isnull().sum()
```
